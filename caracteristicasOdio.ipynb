<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Pregunta 8.</span>
<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Considerando dos grupos de comentarios (odio y no odio) ¿Cuál es el porcentaje de palabras en cada combinación posible de género y número (p.ej. masculino singular) en cada grupo?</span>

```python
# Incluye aquí el código generado para poder responder a tu pregunta
grupo_odio_entidad    = []
grupo_no_odio_entidad = []
total_tokens=0
for i, documento in enumerate(doc):
    #print(f"Comentario {i+1}: {data['CONTENIDO A ANALIZAR'][i]}")
    if data['INTENSIDAD'][i] > 0:
        #Interactuar con cada token del comentario donde el valor de intensidad es mayor a 0 es decir con ODIO
        for token in documento:
            genero = token.morph.get("Gender")  # devuelve lista ['Masc'] o []
            numero = token.morph.get("Number")  # devuelve lista ['Plur'] o []
            #Combinaciones posibles de genero y numero: Masculino singular, Masculino plural, Femenino singular, Femenino plural, Neutro singular, Neutro plural
            #print(f"Token: {token.text}, Género: {genero}, Número: {numero}")
            grupo_odio_entidad.append((token.text, genero, numero))
            total_tokens += 1
            
    else:
        #Interactuar con cada token del comentario donde el valor de intensidad es igual a 0 es decir sin ODIO
        for token in documento:
            genero = token.morph.get("Gender")  # devuelve lista ['Masc'] o []
            numero = token.morph.get("Number")  # devuelve lista ['Plur'] o []
           # print(f"Token: {token.text}, Género: {genero}, Número: {numero}")
            grupo_no_odio_entidad.append((token.text, genero, numero))  
            total_tokens += 1
# Convertir listas a dataframe
df_odio = pd.DataFrame(grupo_odio_entidad, columns=["token", "genero", "numero"])
df_no_odio = pd.DataFrame(grupo_no_odio_entidad, columns=["token", "genero", "numero"])

# Limpiar listas a strings simples (vienen como ['Masc'] o [])
df_odio["genero"]    = df_odio["genero"].apply(lambda x: x[0] if x else "N/A")
df_odio["numero"]    = df_odio["numero"].apply(lambda x: x[0] if x else "N/A")
df_no_odio["genero"] = df_no_odio["genero"].apply(lambda x: x[0] if x else "N/A")
df_no_odio["numero"] = df_no_odio["numero"].apply(lambda x: x[0] if x else "N/A")

# Combinar género y número en una sola columna
df_odio["combinacion"]    = df_odio["genero"]    + " + " + df_odio["numero"]
df_no_odio["combinacion"] = df_no_odio["genero"] + " + " + df_no_odio["numero"]

resumen_completo = pd.DataFrame({
    "CON odio (n)":  df_odio["combinacion"].value_counts(),
    "CON odio (%)":  df_odio["combinacion"].value_counts(normalize=True).mul(100).round(2),
    "SIN odio (n)":  df_no_odio["combinacion"].value_counts(),
    "SIN odio (%)":  df_no_odio["combinacion"].value_counts(normalize=True).mul(100).round(2),
}).fillna(0)

print(f"\nTotal tokens CON odio:    {len(df_odio)}")
print(f"Total tokens SIN odio:    {len(df_no_odio)}")
print(f"Total tokens corpus:      {total_tokens}")
print()
print(resumen_completo)        
```

<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">¿Es posible utilizar alguna de las características extraídas en las preguntas anteriores para determinar si un mensaje contiene odio? Justifica tu respuesta con el análisis estadístico que consideres necesario.</span>

```python
def definir_diccionario_lemas_odio(df_odio, df_no_odio):
    # Aquí puedes definir tu diccionario de lemas relacionados con el odio
    freq_odio    = df_odio["lema"].value_counts(normalize=True).mul(100).round(4)
    freq_no_odio = df_no_odio["lema"].value_counts(normalize=True).mul(100).round(4)

    diccionario_odio = pd.DataFrame({
        "frec_odio (%)":    freq_odio,
        "frec_no_odio (%)": freq_no_odio.reindex(freq_odio.index).fillna(0),
    }).reset_index().rename(columns={"index": "lema"})

    # Ordenar por frecuencia en odio
    diccionario_odio = diccionario_odio.sort_values("frec_odio (%)", ascending=False).head(200)
    diccionario_odio = diccionario_odio[(diccionario_odio["frec_odio (%)"]    > 0.1) 
                                        & (diccionario_odio["frec_no_odio (%)"] < 0.1)     # raro en no odio
                            ]       
    #print(diccionario_odio)
    return diccionario_odio

def clasificar_por_lemas(texto, lemas_odio):
    doc_nuevo = nlp(texto)
    
    lemas_comentario = [
        token.lemma_.lower()
        for token in doc_nuevo
        if not token.is_stop and not token.is_punct
    ]
    
    coincidencias = set(lemas_comentario) & lemas_odio
    
    return {
        "prediccion":     "odio"    if coincidencias else "no odio",
        "coincidencias":  coincidencias,
        "lemas":          lemas_comentario
    }

diccionario_frecuencias_odio=definir_diccionario_lemas_odio(df_odio, df_no_odio)
print(f"Lemas discriminativos de odio: {len(diccionario_frecuencias_odio)}")
print(diccionario_frecuencias_odio)
print(type(diccionario_frecuencias_odio)) 

lemas_odio_set = set(diccionario_frecuencias_odio["lema"].values)

# Probar con ejemplos manuales
ejemplos = [
    "eres un imbécil y una vergüenza",       # esperado: odio
    "el partido estuvo muy bien jugado",      # esperado: no odio
    "este árbitro es un idiota corrupto",     # esperado: odio
    "mañana juegan en el estadio",            # esperado: no odio
]

for texto in ejemplos:
    resultado = clasificar_por_lemas(texto, lemas_odio_set)
    print(f"\nTexto:         {texto}")
    print(f"Predicción:    {resultado['prediccion']}")
    print(f"Coincidencias: {resultado['coincidencias']}")
    print(f"Lemas:         {resultado['lemas']}")

```
